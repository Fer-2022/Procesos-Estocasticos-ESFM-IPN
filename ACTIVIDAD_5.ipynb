<p style="color:#D12B2B; font-size:28px; font-weight:bold;">
  Ejercicio 3: Aproximación de P(t) para una CMTC
</p>

<p style="color:#1A73E8; font-size:22px; font-weight:bold;">
  Problema: Calcular P(0.5), P(1) y P(5) usando la serie infinita aproximada con los primeros M términos.
</p>

<p style="font-size:18px;">
  Usaremos la fórmula:
</p>

<p style="font-size:20px; font-weight:bold;">
  P(t)=Σ e^(-rt) ((rt)^k / k!) P_hat^k
</p>

<p style="font-size:18px;">
  En este ejercicio se usa r = 6 y la matriz R del ejercicio anterior.
</p>

```python
import numpy as np
import math

# Matriz de tasas R del ejercicio anterior
R = np.array([
    [0, 2, 3, 0],
    [4, 0, 2, 0],
    [0, 2, 0, 2],
    [1, 0, 3, 0]
], dtype=float)

# Valor dado de r
r = 6

# Calculamos r_i como la suma de cada fila
r_i = R.sum(axis=1)

print("Matriz R:")
print(R)

print("\nValores de r_i:")
print(r_i)
```

<p style="color:#1A73E8; font-size:22px; font-weight:bold;">
  Construcción de la matriz P_hat
</p>

<p style="font-size:18px;">
  Para construir P_hat usamos:
</p>

<p style="font-size:20px; font-weight:bold;">
  p_hat_ii = 1 - r_i/r
</p>

<p style="font-size:20px; font-weight:bold;">
  p_hat_ij = r_ij/r, cuando i ≠ j
</p>

```python
# Construimos P_hat
n = R.shape[0]
P_hat = np.zeros((n, n))

for i in range(n):
    for j in range(n):
        if i == j:
            P_hat[i, j] = 1 - r_i[i] / r
        else:
            P_hat[i, j] = R[i, j] / r

print("Matriz P_hat:")
print(P_hat)

print("\nComprobación de suma por filas:")
print(P_hat.sum(axis=1))
```

<p style="color:#1A73E8; font-size:22px; font-weight:bold;">
  Elección de M
</p>

<p style="font-size:18px;">
  El problema recomienda usar:
</p>

<p style="font-size:20px; font-weight:bold;">
  M ≈ max{rt + 5√rt, 20}
</p>

<p style="font-size:18px;">
  Como debemos calcular P(0.5), P(1) y P(5), calcularemos M para cada valor de t.
</p>

```python
def calcular_M(t, r):
    return math.ceil(max(r*t + 5*math.sqrt(r*t), 20))

for t in [0.5, 1, 5]:
    print(f"Para t = {t}, M = {calcular_M(t, r)}")
```

<p style="color:#1A73E8; font-size:22px; font-weight:bold;">
  Función para calcular P(t)
</p>

<p style="font-size:18px;">
  Esta función aproxima P(t) sumando los primeros M términos de la serie:
</p>

<p style="font-size:20px; font-weight:bold;">
  P(t)=Σ e^(-rt) ((rt)^k / k!) P_hat^k
</p>

```python
def calcular_P_t(P_hat, r, t):
    n = P_hat.shape[0]
    M = calcular_M(t, r)

    P_t = np.zeros((n, n))
    P_hat_k = np.eye(n)

    for k in range(M + 1):
        coeficiente = math.exp(-r*t) * ((r*t)**k) / math.factorial(k)
        P_t += coeficiente * P_hat_k

        P_hat_k = P_hat_k @ P_hat

    return P_t, M
```

<p style="color:#1A73E8; font-size:22px; font-weight:bold;">
  Cálculo de P(0.5), P(1) y P(5)
</p>

<p style="font-size:18px;">
  Ahora calculamos las matrices de transición para los tres tiempos solicitados.
</p>

```python
tiempos = [0.5, 1, 5]
resultados = {}

for t in tiempos:
    P_t, M = calcular_P_t(P_hat, r, t)
    resultados[t] = P_t

    print(f"\nP({t}) usando M = {M}:")
    print(np.round(P_t, 6))

    print("Suma por filas:")
    print(np.round(P_t.sum(axis=1), 6))
```

<p style="color:#D12B2B; font-size:28px; font-weight:bold;">
  Verificación de Chapman-Kolmogorov
</p>

<p style="color:#1A73E8; font-size:22px; font-weight:bold;">
  Se debe verificar si P(1) = P(0.5)P(0.5)
</p>

<p style="font-size:18px;">
  Calcularemos P(0.5)P(0.5) y lo compararemos con P(1).
</p>

```python
P_05 = resultados[0.5]
P_1 = resultados[1]

producto = P_05 @ P_05

print("P(1):")
print(np.round(P_1, 6))

print("\nP(0.5)P(0.5):")
print(np.round(producto, 6))

print("\nDiferencia P(1) - P(0.5)P(0.5):")
print(np.round(P_1 - producto, 10))
```

<p style="color:#1A73E8; font-size:22px; font-weight:bold;">
  Conclusión
</p>

<p style="font-size:18px;">
  Si la diferencia entre P(1) y P(0.5)P(0.5) es muy cercana a cero, entonces se verifica la ecuación de Chapman-Kolmogorov.
</p>

```python
diferencia = np.linalg.norm(P_1 - producto)

print("Norma de la diferencia:")
print(diferencia)

if diferencia < 1e-6:
    print("\nConclusión: Sí se verifica Chapman-Kolmogorov.")
else:
    print("\nConclusión: No se verifica exactamente, pero la diferencia se debe a la aproximación numérica.")
```

<p style="font-size:18px;">
  Conclusión: Sí se verifica la ecuación de Chapman-Kolmogorov, ya que la matriz P(1) es prácticamente igual al producto P(0.5)P(0.5). Las pequeñas diferencias que puedan aparecer se deben a la aproximación numérica de la serie infinita.
</p>

<p style="color:#D12B2B; font-size:28px; font-weight:bold;">
  Ejercicio 4.2: Algoritmo de uniformización para calcular P(t)
</p>

<p style="color:#1A73E8; font-size:22px; font-weight:bold;">
  Problema: Repetir el ejercicio 3 usando el algoritmo de uniformización con tolerancia ε = 0.00001.
</p>

<p style="font-size:18px;">
  El objetivo es calcular P(0.5), P(1) y P(5), indicando el valor de M utilizado en cada caso.
</p>

```python
import numpy as np
import math
```

<p style="color:#1A73E8; font-size:22px; font-weight:bold;">
  Datos del problema
</p>

<p style="font-size:18px;">
  Usaremos la matriz de tasas R del ejercicio anterior y el valor r = 6.
</p>

```python
R = np.array([
    [0, 2, 3, 0],
    [4, 0, 2, 0],
    [0, 2, 0, 2],
    [1, 0, 3, 0]
], dtype=float)

r = 6
epsilon = 0.00001

print("Matriz R:")
print(R)

print("\nr =", r)
print("epsilon =", epsilon)
```

<p style="color:#1A73E8; font-size:22px; font-weight:bold;">
  Construcción de la matriz P_hat
</p>

<p style="font-size:18px;">
  Primero calculamos r_i como la suma de cada fila de R.
</p>

<p style="font-size:18px;">
  Después usamos:
</p>

<p style="font-size:20px; font-weight:bold;">
  p_hat_ii = 1 - r_i/r
</p>

<p style="font-size:20px; font-weight:bold;">
  p_hat_ij = r_ij/r, si i ≠ j
</p>

```python
r_i = R.sum(axis=1)

n = R.shape[0]
P_hat = np.zeros((n, n))

for i in range(n):
    for j in range(n):
        if i == j:
            P_hat[i, j] = 1 - r_i[i] / r
        else:
            P_hat[i, j] = R[i, j] / r

print("Valores de r_i:")
print(r_i)

print("\nMatriz P_hat:")
print(P_hat)

print("\nSuma de filas de P_hat:")
print(P_hat.sum(axis=1))
```

<p style="color:#1A73E8; font-size:22px; font-weight:bold;">
  Algoritmo de uniformización
</p>

<p style="font-size:18px;">
  El algoritmo suma términos de la serie hasta que la suma acumulada de probabilidades de Poisson sea mayor o igual que 1 - ε.
</p>

<p style="font-size:18px;">
  Esto garantiza que el error de aproximación sea menor o igual que ε.
</p>

```python
def uniformizacion(P_hat, r, t, epsilon):
    n = P_hat.shape[0]

    # Paso 1
    k = 0
    c = math.exp(-r * t)
    suma = c

    # Paso 2
    A = np.eye(n)

    # Paso 3
    B = c * A

    # Paso 4 y 5
    while suma < 1 - epsilon:
        k = k + 1

        c = c * (r * t) / k
        A = A @ P_hat
        B = B + c * A

        suma = suma + c

    M = k
    return B, M, suma
```

<p style="color:#1A73E8; font-size:22px; font-weight:bold;">
  Cálculo de P(0.5), P(1) y P(5)
</p>

<p style="font-size:18px;">
  Ahora aplicamos el algoritmo de uniformización para cada valor de t.
</p>

```python
tiempos = [0.5, 1, 5]
resultados_uniformizacion = {}

for t in tiempos:
    P_t, M, suma = uniformizacion(P_hat, r, t, epsilon)
    resultados_uniformizacion[t] = P_t

    print(f"\nP({t}) con tolerancia epsilon = {epsilon}")
    print(f"Valor de M usado: {M}")
    print(f"Suma acumulada de probabilidades de Poisson: {suma}")

    print("\nMatriz P(t):")
    print(np.round(P_t, 6))

    print("\nSuma por filas:")
    print(np.round(P_t.sum(axis=1), 6))
```

<p style="color:#D12B2B; font-size:28px; font-weight:bold;">
  Verificación de Chapman-Kolmogorov
</p>

<p style="color:#1A73E8; font-size:22px; font-weight:bold;">
  Se verifica si P(1) = P(0.5)P(0.5)
</p>

```python
P_05 = resultados_uniformizacion[0.5]
P_1 = resultados_uniformizacion[1]

producto = P_05 @ P_05

print("P(1):")
print(np.round(P_1, 6))

print("\nP(0.5)P(0.5):")
print(np.round(producto, 6))

print("\nDiferencia P(1) - P(0.5)P(0.5):")
print(np.round(P_1 - producto, 10))

norma = np.linalg.norm(P_1 - producto)

print("\nNorma de la diferencia:")
print(norma)
```

<p style="color:#1A73E8; font-size:22px; font-weight:bold;">
  Conclusión de Chapman-Kolmogorov
</p>

```python
if norma < 1e-4:
    print("Conclusión: Sí se verifica Chapman-Kolmogorov de forma aproximada.")
else:
    print("Conclusión: No se verifica exactamente, pero la diferencia se debe a la aproximación numérica.")
```

<p style="color:#D12B2B; font-size:28px; font-weight:bold;">
  Comparación con el ejercicio 3
</p>

<p style="font-size:18px;">
  En el ejercicio 3 se eligió M con la regla aproximada M ≈ max{rt + 5√rt, 20}.
</p>

<p style="font-size:18px;">
  Ahora el valor de M se obtiene automáticamente con la tolerancia ε = 0.00001.
</p>

```python
def M_ejercicio_3(t, r):
    return math.ceil(max(r*t + 5*math.sqrt(r*t), 20))

print("Comparación de valores de M:")

for t in tiempos:
    M_regla = M_ejercicio_3(t, r)
    _, M_uniformizacion, _ = uniformizacion(P_hat, r, t, epsilon)

    print(f"\nt = {t}")
    print(f"M del ejercicio 3: {M_regla}")
    print(f"M por uniformización con epsilon = {epsilon}: {M_uniformizacion}")
```

<p style="color:#1A73E8; font-size:22px; font-weight:bold;">
  Conclusión final
</p>

<p style="font-size:18px;">
  El algoritmo de uniformización calcula P(t) sumando términos hasta que la cola de la distribución Poisson sea menor que ε.
</p>

<p style="font-size:18px;">
  Por eso, el valor de M no se elige de forma fija, sino que depende de t, de r y de la tolerancia solicitada.
</p>

<p style="font-size:18px;">
  Con ε = 0.00001, las matrices obtenidas son aproximaciones confiables de P(0.5), P(1) y P(5).
