<p style="color:#D12B2B; font-size:26px; font-weight:bold;">
  Punto 1: Calcular P(t) en SymPy
</p>

<p style="color:#000000; font-size:18px;">
  Se tiene una matriz generadora Q de una cadena de Markov en tiempo continuo.
  Esta matriz no es estocástica, porque sus filas no tienen que sumar 1.
  En una matriz generadora, las filas deben sumar 0.
</p>

<p style="color:#D12B2B; font-size:22px; font-weight:bold;">
  Matriz generadora Q:
</p>

<p style="color:#000000; font-size:20px;">
\[
Q=
\begin{pmatrix}
-2 & 2 & 0\\
3 & -4 & 1\\
0.5 & 0 & -0.5
\end{pmatrix}
\]
</p>

<p style="color:#1A73E8; font-size:20px;">
  Para obtener la matriz de probabilidades de transición se calcula:
</p>

<p style="color:#000000; font-size:22px;">
\[
P(t)=e^{tQ}
\]
</p>

<p style="color:#000000; font-size:18px;">
  En Python usaremos SymPy para definir la variable t, construir la matriz Q
  y después calcular la exponencial matricial.
</p>

```python
# Punto 1: Calcular P(t) usando SymPy

import sympy as sp

# Definimos la variable simbólica t
t = sp.symbols('t', real=True, nonnegative=True)

# Definimos la matriz generadora Q
Q = sp.Matrix([
    [-2,  2,  0],
    [ 3, -4,  1],
    [sp.Rational(1, 2), 0, -sp.Rational(1, 2)]
])

# Mostramos Q
print("Matriz generadora Q:")
display(Q)

# Verificamos que las filas suman cero
print("Suma de las filas de Q:")
display([sum(Q.row(i)) for i in range(Q.rows)])

# Calculamos P(t) = e^(tQ)
P_t = sp.exp(t * Q)

# Simplificamos la matriz
P_t = sp.simplify(P_t)

print("Matriz de transición P(t) = e^(tQ):")
display(P_t)
```

<p style="color:#D12B2B; font-size:26px; font-weight:bold;">
  Punto 2: Probabilidad de estar caído después de t = 1
</p>

<p style="color:#000000; font-size:18px;">
  La pregunta pide calcular la probabilidad de que el sistema esté en el estado caído después de una unidad de tiempo.
</p>

<p style="color:#000000; font-size:18px;">
  Los estados del sistema son:
</p>

<p style="color:#000000; font-size:18px;">
  Estado 0: servidor libre <br>
  Estado 1: servidor ocupado <br>
  Estado 2: servidor caído
</p>

<p style="color:#1A73E8; font-size:20px; font-weight:bold;">
  Condición inicial
</p>

<p style="color:#000000; font-size:18px;">
  Si el sistema inicia libre, entonces inicia en el estado 0.
  Por lo tanto, la probabilidad que se busca es:
</p>

<p style="color:#000000; font-size:22px;">
\[
P_{02}(1)
\]
</p>

<p style="color:#000000; font-size:18px;">
  Esto significa: probabilidad de pasar del estado 0 al estado 2 en tiempo t = 1.
</p>

<p style="color:#000000; font-size:18px;">
  Para obtenerla, primero calculamos:
</p>

<p style="color:#000000; font-size:22px;">
\[
P(t)=e^{tQ}
\]
</p>

<p style="color:#000000; font-size:18px;">
  Después sustituimos t = 1 y tomamos la entrada de la fila 0, columna 2.
</p>

```python
import sympy as sp

# Definimos la variable simbólica
t = sp.symbols('t', real=True, nonnegative=True)

# Matriz generadora Q
Q = sp.Matrix([
    [-2,  2,  0],
    [ 3, -4,  1],
    [sp.Rational(1, 2), 0, -sp.Rational(1, 2)]
])

# Calculamos P(t)
P_t = sp.exp(t * Q)
P_t = sp.simplify(P_t)

# Evaluamos en t = 1
P_1 = P_t.subs(t, 1)

# Probabilidad de iniciar libre, estado 0, y terminar caído, estado 2
probabilidad_caido = P_1[0, 2]

print("Matriz P(1):")
display(P_1)

print("Probabilidad de estar caído después de t = 1, iniciando libre:")
display(probabilidad_caido)

print("Valor decimal:")
display(sp.N(probabilidad_caido, 6))
```

<p style="color:#D12B2B; font-size:26px; font-weight:bold;">
  Punto 2: Comportamiento de P(t) cuando t → ∞
</p>

<p style="color:#000000; font-size:18px;">
  La matriz P(t) representa las probabilidades de transición entre los estados del sistema conforme pasa el tiempo.
</p>

<p style="color:#000000; font-size:18px;">
  Cuando t crece mucho, es decir, cuando t → ∞, la cadena de Markov tiende a estabilizarse.
  Esto significa que las probabilidades dejan de cambiar de forma importante y se acercan a una distribución estacionaria.
</p>

<p style="color:#1A73E8; font-size:20px; font-weight:bold;">
  Distribución estacionaria
</p>

<p style="color:#000000; font-size:18px;">
  La distribución estacionaria se representa como:
</p>

<p style="color:#000000; font-size:22px;">
\[
\pi = (\pi_0,\pi_1,\pi_2)
\]
</p>

<p style="color:#000000; font-size:18px;">
  Y debe cumplir:
</p>

<p style="color:#000000; font-size:22px;">
\[
\pi Q = 0
\]
</p>

<p style="color:#000000; font-size:18px;">
  Además, como son probabilidades, debe cumplirse que:
</p>

<p style="color:#000000; font-size:22px;">
\[
\pi_0+\pi_1+\pi_2=1
\]
</p>

<p style="color:#000000; font-size:18px;">
  Entonces, al calcular el límite de P(t), todas las filas de la matriz se acercan a la misma distribución estacionaria.
</p>

```python
import sympy as sp

# Definimos la variable simbólica
t = sp.symbols('t', real=True, nonnegative=True)

# Matriz generadora Q
Q = sp.Matrix([
    [-2,  2,  0],
    [ 3, -4,  1],
    [sp.Rational(1, 2), 0, -sp.Rational(1, 2)]
])

# Calculamos P(t)
P_t = sp.exp(t * Q)
P_t = sp.simplify(P_t)

print("Matriz P(t):")
display(P_t)

# Calculamos el límite cuando t tiende a infinito
P_infinito = P_t.applyfunc(lambda x: sp.limit(x, t, sp.oo))

print("Límite de P(t) cuando t tiende a infinito:")
display(P_infinito)

# Calculamos la distribución estacionaria pi
pi0, pi1, pi2 = sp.symbols('pi0 pi1 pi2')

pi = sp.Matrix([[pi0, pi1, pi2]])

# Sistema: pi Q = 0 y pi0 + pi1 + pi2 = 1
ecuaciones = list(pi * Q) + [pi0 + pi1 + pi2 - 1]

solucion_pi = sp.solve(ecuaciones, (pi0, pi1, pi2), dict=True)

print("Distribución estacionaria:")
display(solucion_pi)
```

<p style="color:#D12B2B; font-size:26px; font-weight:bold;">
  Punto 3: Graficar P00(t), P01(t) y P02(t)
</p>

<p style="color:#000000; font-size:18px;">
  Como el sistema inicia libre, se toma la fila 0 de la matriz P(t).
  Esa fila representa las probabilidades de pasar del estado 0 hacia cada uno de los estados posibles.
</p>

<p style="color:#000000; font-size:18px;">
  Por lo tanto:
</p>

<p style="color:#000000; font-size:20px;">
  P<sub>00</sub>(t): probabilidad de seguir libre en el tiempo t. <br>
  P<sub>01</sub>(t): probabilidad de pasar de libre a ocupado en el tiempo t. <br>
  P<sub>02</sub>(t): probabilidad de pasar de libre a caído en el tiempo t.
</p>

<p style="color:#1A73E8; font-size:20px; font-weight:bold;">
  Objetivo de la gráfica
</p>

<p style="color:#000000; font-size:18px;">
  La gráfica permite observar cómo cambian las probabilidades conforme avanza el tiempo.
  Al inicio, cuando t = 0, el sistema está totalmente en el estado libre, por eso P<sub>00</sub>(0) = 1.
</p>

```python
import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

# Variable simbólica
t = sp.symbols('t', real=True, nonnegative=True)

# Matriz generadora Q
Q = sp.Matrix([
    [-2,  2,  0],
    [ 3, -4,  1],
    [sp.Rational(1, 2), 0, -sp.Rational(1, 2)]
])

# Matriz de transición P(t)
P_t = sp.exp(t * Q)
P_t = sp.simplify(P_t)

# Extraemos las funciones de la fila 0
P00 = P_t[0, 0]
P01 = P_t[0, 1]
P02 = P_t[0, 2]

print("P00(t):")
display(P00)

print("P01(t):")
display(P01)

print("P02(t):")
display(P02)

# Convertimos las expresiones simbólicas a funciones numéricas
P00_func = sp.lambdify(t, P00, 'numpy')
P01_func = sp.lambdify(t, P01, 'numpy')
P02_func = sp.lambdify(t, P02, 'numpy')

# Valores de tiempo
tiempo = np.linspace(0, 10, 300)

# Evaluamos las probabilidades
valores_P00 = P00_func(tiempo)
valores_P01 = P01_func(tiempo)
valores_P02 = P02_func(tiempo)

# Graficamos
plt.figure(figsize=(10, 6))

plt.plot(tiempo, valores_P00, label=r'$P_{00}(t)$')
plt.plot(tiempo, valores_P01, label=r'$P_{01}(t)$')
plt.plot(tiempo, valores_P02, label=r'$P_{02}(t)$')

plt.title('Probabilidades de transición iniciando en estado libre')
plt.xlabel('Tiempo t')
plt.ylabel('Probabilidad')
plt.grid(True)
plt.legend()
plt.show()
```


<p style="color:#D12B2B; font-size:26px; font-weight:bold;">
  Punto 4: Simulación de trayectorias
</p>

<p style="color:#000000; font-size:18px;">
  En este punto se simulan varias trayectorias de la cadena de Markov en tiempo continuo.
  Una trayectoria representa cómo cambia el estado del sistema conforme pasa el tiempo.
</p>

<p style="color:#000000; font-size:18px;">
  Los estados son:
</p>

<p style="color:#000000; font-size:20px;">
  Estado 0: servidor libre <br>
  Estado 1: servidor ocupado <br>
  Estado 2: servidor caído
</p>

<p style="color:#1A73E8; font-size:20px; font-weight:bold;">
  Idea de la simulación
</p>

<p style="color:#000000; font-size:18px;">
  La matriz generadora Q indica las tasas de cambio entre estados.
  Los elementos fuera de la diagonal representan las tasas de transición.
  La diagonal negativa representa la tasa total de salida de cada estado.
</p>

<p style="color:#000000; font-size:18px;">
  Para simular una trayectoria, se realiza lo siguiente:
</p>

<p style="color:#000000; font-size:18px;">
  1. Se inicia en el estado 0, es decir, servidor libre. <br>
  2. Se calcula cuánto tiempo permanece el sistema en ese estado. <br>
  3. Se elige el siguiente estado con base en las tasas de transición. <br>
  4. Se repite el proceso hasta llegar al tiempo final.
</p>

```python
import numpy as np
import matplotlib.pyplot as plt
import sympy as sp

# Matriz generadora Q en formato NumPy
Q = np.array([
    [-2,   2,   0],
    [ 3,  -4,   1],
    [0.5,  0,  -0.5]
], dtype=float)

# Nombres de los estados
estados = {
    0: "Libre",
    1: "Ocupado",
    2: "Caído"
}

# Función para simular una trayectoria de una CTMC
def simular_trayectoria(Q, estado_inicial=0, tiempo_final=10):
    tiempos = [0]
    trayectoria = [estado_inicial]

    tiempo_actual = 0
    estado_actual = estado_inicial

    while tiempo_actual < tiempo_final:
        # Tasa de salida del estado actual
        tasa_salida = -Q[estado_actual, estado_actual]

        # Si la tasa de salida es cero, se queda en ese estado
        if tasa_salida <= 0:
            break

        # Tiempo de permanencia en el estado actual
        tiempo_espera = np.random.exponential(1 / tasa_salida)
        tiempo_actual += tiempo_espera

        if tiempo_actual > tiempo_final:
            tiempos.append(tiempo_final)
            trayectoria.append(estado_actual)
            break

        # Probabilidades de transición hacia otros estados
        probabilidades = Q[estado_actual, :].copy()
        probabilidades[estado_actual] = 0
        probabilidades = probabilidades / tasa_salida

        # Elegimos el siguiente estado
        siguiente_estado = np.random.choice([0, 1, 2], p=probabilidades)

        tiempos.append(tiempo_actual)
        trayectoria.append(siguiente_estado)

        estado_actual = siguiente_estado

    return tiempos, trayectoria

# Simulamos varias trayectorias
np.random.seed(123)

numero_trayectorias = 5
tiempo_final = 10

plt.figure(figsize=(10, 6))

for i in range(numero_trayectorias):
    tiempos, trayectoria = simular_trayectoria(Q, estado_inicial=0, tiempo_final=tiempo_final)
    plt.step(tiempos, trayectoria, where='post', label=f'Trayectoria {i+1}')

plt.yticks([0, 1, 2], ["Libre", "Ocupado", "Caído"])
plt.xlabel("Tiempo t")
plt.ylabel("Estado del sistema")
plt.title("Simulación de trayectorias de la cadena de Markov")
plt.grid(True)
plt.legend()
plt.show()
```
